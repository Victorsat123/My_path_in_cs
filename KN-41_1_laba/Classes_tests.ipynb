### Перевірка даних ###
1. assert перевіряє твердження та зупиняє програму, якщо умова хибна:


```python
number = -1
assert number > 0, "Число має бути більшим за нуль!"
```

2. raise ValueError доречний, коли програма має повідомити про неправильний аргумент:

```python
def create_name(name):
    if not name:
        raise ValueError("Ім'я не може бути порожнім")
    return name
print(create_name("Віктор"))
```

```python
def create_name(name):
    if not name:
        raise ValueError("Ім'я не може бути порожнім")
    return name

print(create_name(""))
```

### Тестування функцій ###


1.Юніт-тест можна написати для окремої функції. Спочатку визначте очікуваний результат для звичайних, порожніх і граничних даних:


```python
def count_symbols(text: str) -> int:
    return len(text)
```

2.Перевірте функцію за допомогою unittest (тест має бути у окремому класі, який наслідує unittest.TestCase):


```python
def test_count_symbols(self):
    self.assertEqual(3, count_symbols("abc"))
    self.assertEqual(0, count_symbols(""))
```
