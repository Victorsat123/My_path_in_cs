### Перевірка даних ###
#### 3.Створіть власну перевірку введеного числа та опишіть у звіті результат для правильного й неправильного введення: ####

```python
def validate_number(value_str: str) -> float:
    try:
        number = float(value_str)
    except ValueError:
        raise ValueError("Введене значення не є дійсним числом")

    if number <= 0:
        raise ValueError("Число має бути суворо більшим за 0")

    return number


try:
    user_input = input("Введіть додатне число: ")
    valid_number = validate_number(user_input)
    print(f"Успішно: прийнято число {valid_number}")
except ValueError as error:
    print(f"Помилка валідації: {error}")
```

### Тестування функцій ###
#### 4-5.Створіть функцію, яка рахує кількість голосних літер у рядку, і напишіть щонайменше три тести для неї. ####

```python
import unittest


def count_vowels(text: str) -> int:
    vowels = "аеєиіїоуюяaeiouy"
    return sum(1 for char in text.lower() if char in vowels)


class TestCountVowels(unittest.TestCase):

    # Завдання 4: базові перевірки
    def test_standard_words(self):
        self.assertEqual(count_vowels("hello"), 2)
        self.assertEqual(count_vowels("python"), 2)

    def test_mixed_case(self):
        self.assertEqual(count_vowels("PyThOn"), 2)
        self.assertEqual(count_vowels("HELLO"), 2)

    def test_no_vowels(self):
        self.assertEqual(count_vowels("bcdfg"), 0)

    # Завдання 5: крайові випадки
    def test_empty_string(self):
        self.assertEqual(count_vowels(""), 0)

    def test_digits_and_symbols(self):
        self.assertEqual(count_vowels("12345!@#"), 0)
        self.assertEqual(count_vowels("lab2026"), 1)

    def test_ukrainian_letters(self):
        self.assertEqual(count_vowels("привіт"), 2)
        self.assertEqual(count_vowels("програмування"), 5)
        self.assertEqual(count_vowels("Єнот"), 2)


# Запуск у Jupyter Notebook
if __name__ == "__main__":
    unittest.main(argv=[""], exit=False)
```
